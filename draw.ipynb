```python
import matplotlib.pyplot as plt

sizes = [100, 1000, 5000, 10000]
seq_naive = [0.000276, 0.188956, 31.57629, 276.532246]
cpu_naive = [0.021383, 0.083086, 10.16129, 79.17651]
gpu_naive = [0.353139, 0.046220, 3.66277, 22.200161]
seq_strassen = [0.000855, 0.19223, 21.810084, 179.333068]
cpu_strassen = [0.077396, 0.22734, 5.41947, 62.94149]
gpu_strassen = [0.041739, 0.38153, 3.150007, 25.231380]

plt.figure(figsize=(10, 6))
plt.plot(sizes, seq_naive, marker='o', label='Seq Naive')
plt.plot(sizes, cpu_naive, marker='o', label='CPU Naive')
plt.plot(sizes, gpu_naive, marker='o', label='GPU Naive')
plt.plot(sizes, seq_strassen, marker='o', label='Seq Strassen')
plt.plot(sizes, cpu_strassen, marker='o', label='CPU Strassen')
plt.plot(sizes, gpu_strassen, marker='o', label='GPU Strassen')

plt.xlabel('Size')
plt.ylabel('Time (seconds)')
plt.title('Trực quan hóa thời gian thực thi các phiên bản thuật toán nhân ma trận theo kích thước ma trận')
plt.legend()
plt.yscale('log')
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.show()
```
